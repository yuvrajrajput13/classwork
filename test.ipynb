```python
import pandas as pd
df=pd.read_csv("/content/titanic.csv")
print(df)
```

```python
age_median = df['Age'].median()
df['Age'] = df['Age'].fillna(age_median)

embarked_mode = df['Embarked'].mode()[0]
df['Embarked'] = df['Embarked'].fillna(embarked_mode)

df.drop('Cabin', axis=1, inplace=True, errors='ignore')

print("Missing values after imputation and dropping 'Cabin' column:")
print(df.isnull().sum())
```

```python
age_median = df['Age'].median()
df['Age'] = df['Age'].fillna(age_median)

embarked_mode = df['Embarked'].mode()[0]
df['Embarked'] = df['Embarked'].fillna(embarked_mode)

print("Missing values after imputation:")
print(df.isnull().sum())
```

```python
df = pd.get_dummies(df, columns=['Sex', 'Embarked'], drop_first=True)
print(df.head())
```

```python
import pandas as pd

columns_to_encode = [col for col in ['Sex', 'Embarked'] if col in df.columns]

if columns_to_encode:
    df = pd.get_dummies(df, columns=columns_to_encode, drop_first=True)
    print(f"Columns {columns_to_encode} were one-hot encoded.")
elif not columns_to_encode and ('Sex_male' in df.columns or 'Embarked_Q' in df.columns):
    print("Columns 'Sex' and 'Embarked' have already been one-hot encoded.")
else:
    print("Columns 'Sex' and 'Embarked' do not exist in the DataFrame and were not encoded.")

print(df.head())
```
